```python
# series创建
import pandas as pd
s = pd.Series([1,2,3,4,5,6])
print(s)
# 自定义索引
# 定义name
s = pd.Series([1,2,3,4,5,6],index=['a','b','c','d','e','f'],name='月')
print(s)
```

```python
# 通过字典方式创建
s = pd.Series({'a':1,'b':2,'c':3})
print(s)
# 基于原series创建
s1 = pd.Series(s,index=['a','c'])
print(s1)
```

```python
# series的属性
'''
index: Series的索引对象
values: Series的值
dtype dtypes: 元素类型
shape: 形状
ndim: 维度
size: 元素个数
name: 名称
loc[]: 显式索引, 按标签索引或切片
iloc[]: 隐式索引, 按位置索引或切片
at[]: 使用标签访问单个元素
iat[]: 使用位置访问单个元素
'''
```

```python
import pandas as pd
s = pd.Series([1,2,3,4,5,6],index=['a','b','c','d','e','f'],name='月')
```

```python
print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
print(s.dtype,s.name)
```

```python
print(s.loc['a']) # 显式索引
print(s.loc['a':'c']) # 切片 包括a和c
print(s.iloc[0]) # 隐式索引
print(s.iloc[0:2]) # 包括0不包括2
```

```python
print(s.at['a']) # 显式索引 单个元素
print(s.iat[0]) # 隐式索引 单个元素
```

```python
# 访问数据
print(s['a']) # 显式
print(s[s<3]) # 布尔索引
```

```python
print(s.head()) # 默认前五行, 可自定义数值
print(s.tail(3))
```
